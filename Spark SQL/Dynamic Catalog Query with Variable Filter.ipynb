# Dynamic Catalog Query with Variable Filter

Parses the current username to derive the catalog name, then queries `ca_opportunities` in the `retail` schema using a session variable filter.

```sql
-- Parse the current username to extract the catalog name (text before @)
DECLARE OR REPLACE VARIABLE catalog_name STRING;
SET VARIABLE catalog_name = substring_index(current_user(), '@', 1);

-- Use EXECUTE IMMEDIATE to dynamically set the catalog
EXECUTE IMMEDIATE CONCAT('USE CATALOG ', catalog_name);
USE SCHEMA retail;

-- Verify the active catalog and schema
SELECT catalog_name AS derived_catalog, current_catalog() AS active_catalog, current_schema() AS active_schema;
```

```sql
-- Create a session variable to use as a filter
DECLARE OR REPLACE VARIABLE state_filter STRING DEFAULT 'NSW';

SELECT state_filter AS filter_value;
```

```sql

-- Use EXECUTE IMMEDIATE with the variable to filter ca_opportunities by province
EXECUTE IMMEDIATE 
  CONCAT('SELECT * FROM ', catalog_name, '.retail.au_opportunities WHERE province = ?')
  WHERE state = state_filter;
```

```sql
-- Use the session variable with : syntax to filter ca_opportunities by province
SELECT * FROM aus_opportunities WHERE state = :state_filter;
```
