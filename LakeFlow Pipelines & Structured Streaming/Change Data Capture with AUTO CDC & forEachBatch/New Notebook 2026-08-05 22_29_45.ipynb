```sql
DECLARE OR REPLACE VARIABLE catalog STRING DEFAULT split(current_user(), '@')[0]
```

```sql

USE CATALOG IDENTIFIER(catalog);
USE SCHEMA sdp_2_silver
```

```sql
DROP TABLE IF EXISTS foreachbatch_target;
CREATE TABLE foreachbatch_target 
(
  address STRING COLLATE UTF8_BINARY,
  city STRING COLLATE UTF8_BINARY,
  customer_id BIGINT,
  email STRING COLLATE UTF8_BINARY,
  name STRING COLLATE UTF8_BINARY,
  state STRING COLLATE UTF8_BINARY,
  zip_code STRING COLLATE UTF8_BINARY,
  processing_time TIMESTAMP,
  source_file STRING COLLATE UTF8_BINARY,
  timestamp_datetime TIMESTAMP)
USING delta
TBLPROPERTIES (
  'delta.enableDeletionVectors' = 'true',
  'delta.feature.appendOnly' = 'supported',
  'delta.feature.deletionVectors' = 'supported',
  'delta.feature.invariants' = 'supported',
  'delta.minReaderVersion' = '3',
  'delta.minWriterVersion' = '7',
  'delta.parquet.compression.codec' = 'zstd')
```
